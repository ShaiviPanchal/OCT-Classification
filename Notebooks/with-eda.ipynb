### Import necessary libraries

```python
import tensorflow as tf
from tensorflow import keras

import os
import random
import cv2

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn.metrics

# discontinued import tensorflow_addons as tfa
from tensorflow.keras.models import Model
from tensorflow.keras.preprocessing.image import ImageDataGenerator, load_img
from tensorflow.keras.models import load_model
```

```python
base_dir = r"D:\ML_OCT_RS\Dataset"
print('Base directory --> ', os.listdir(base_dir))
```

### Train, Validation and Test Sets

```python
# NOT EXCHANGED

train_dir = r"D:\ML_OCT_RS\Dataset\train"
print("Train Directory --> ", os.listdir(train_dir))

validation_dir = r"D:\ML_OCT_RS\Dataset\val"
print("Validation Directory --> ", os.listdir(validation_dir))

test_dir = r"D:\ML_OCT_RS\Dataset\test"
print("Test Directory --> ", os.listdir(test_dir))
```

```python
cnv = len(os.listdir(os.path.join(train_dir + r"\CNV")))
dme = len(os.listdir(os.path.join(train_dir + r"\DME")))
drusen = len(os.listdir(os.path.join(train_dir + r"\DRUSEN")))
normal = len(os.listdir(os.path.join(train_dir + r"\NORMAL")))

counts = [cnv, dme, drusen, normal]
```

### Class-wise Image Count in Training Data

```python
import os

train_dir = train_dir

class_counts = {}
for class_name in os.listdir(train_dir):
    class_path = os.path.join(train_dir, class_name)
    if os.path.isdir(class_path):
        class_counts[class_name] = len(os.listdir(class_path))

for class_name, count in class_counts.items():
    print(f"{class_name}: {count} images")
```

### Class Weights

```python
from sklearn.utils import class_weight
import numpy as np

class_counts = np.array(counts)
class_labels = np.array([0, 1, 2, 3])  # [CNV -> 0, DME -> 1, DRUSEN -> 2, NORMAL -> 3]

# Calculate class weights
weights = class_weight.compute_class_weight(class_weight='balanced', 
                                            classes=class_labels, 
                                            y=np.repeat(class_labels, class_counts))
class_weights = dict(enumerate(weights))

print("Class Weights:", class_weights)
```

### Test Data Generator

```python
img_size = (224, 224)
batch_size = 32


test_dataset = tf.keras.preprocessing.image_dataset_from_directory(
    test_dir,
    labels='inferred',
    label_mode='categorical',
    class_names=['CNV', 'DME', 'DRUSEN', 'NORMAL'],
    image_size=img_size,
    batch_size=batch_size,
    shuffle=False
)
```

### Train and Validation Generator

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rescale=1./255
)

train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224, 224),   
    batch_size=500,
    class_mode='categorical'
)

val_datagen = ImageDataGenerator(rescale=1./255)

validation_generator = val_datagen.flow_from_directory(
    validation_dir,
    target_size=(224, 224),
    batch_size=16,
    class_mode='categorical'
)
```

### Image Quality & Variability Analysis

```python
img_shapes = []
cls_dirs = ['CNV', 'DME', 'DRUSEN', 'NORMAL']
dataset_path = train_dir  

for cls in cls_dirs:
    class_path = os.path.join(dataset_path, cls)
    if not os.path.exists(class_path):  
        print(f"Directory not found: {class_path}")
        continue  
    
    for img_name in os.listdir(class_path)[:10]:  # Checking a few images per class
        img = cv2.imread(os.path.join(class_path, img_name))
        if img is not None:
            img_shapes.append(img.shape[:2])  # Store (height, width)
        else:
            print(f"Error reading image: {img_name} in {class_path}")

print(set(img_shapes)) 

```

### Feature Exploration: t-SNE Visualization

```python
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE
from tqdm import tqdm

# Check and update the correct dataset path
dataset_path = train_dir
class_labels = ['CNV', 'DME', 'DRUSEN', 'NORMAL'] 
img_size = (224, 224)
num_samples = 500

# To verify that class directories exist
if not os.path.exists(dataset_path):
    raise ValueError(f"Dataset path {dataset_path} does not exist!")

X, y = [], []

for label, cls in enumerate(class_labels):
    class_path = os.path.join(dataset_path, cls)

    if not os.path.exists(class_path):
        print(f"⚠️ Warning: Class folder '{cls}' not found in {dataset_path}. Skipping...")
        continue

    img_files = os.listdir(class_path)[:num_samples]  # Limit samples per class

    for img_file in tqdm(img_files, desc=f"Loading {cls} images"):
        img = cv2.imread(os.path.join(class_path, img_file), cv2.IMREAD_GRAYSCALE)
        img = cv2.resize(img, img_size)
        X.append(img.flatten())
        y.append(label)

X = np.array(X)
y = np.array(y)

# PCA & t-SNE
pca = PCA(n_components=50).fit_transform(X)
tsne = TSNE(n_components=2, perplexity=30, random_state=42).fit_transform(pca)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(tsne[:, 0], tsne[:, 1], c=y, cmap='viridis', alpha=0.7)
plt.colorbar(scatter, ticks=range(len(class_labels)), label="Class")
plt.title("t-SNE Visualization of OCT Dataset")
plt.xlabel("t-SNE Component 1")
plt.ylabel("t-SNE Component 2")
plt.show()
```

### Sample Visualizations

```python
# Set number of images per class
num_samples = 5 

# Plot Sample Images
fig, axes = plt.subplots(len(class_labels), num_samples, figsize=(12, 8))

for i, cls in enumerate(class_labels):
    class_path = os.path.join(dataset_path, cls)
    img_files = os.listdir(class_path)

    # Select random images from each class
    selected_images = random.sample(img_files, num_samples)

    for j, img_file in enumerate(selected_images):
        img_path = os.path.join(class_path, img_file)
        img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)  
        img = cv2.resize(img, img_size)

        axes[i, j].imshow(img, cmap='gray')  
        axes[i, j].axis("off")
        axes[i, j].set_title(cls, fontsize=10)  # Add class name as title


plt.suptitle("Sample OCT Images from Each Class", fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

### Defining Pre-trained model - InceptionV3

```python
from tensorflow.keras.applications import InceptionV3
inc = InceptionV3(input_shape=(224,224,3),weights='imagenet',include_top=False)
for i in inc.layers:
    i.trainable = False
print(inc.summary())
```

```python
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense, Dropout, Flatten

model = tf.keras.models.Sequential([
    inc,
    Flatten(),
    tf.keras.layers.Dense(4, activation = 'softmax')
])
```

### Defining Loss Function, Optimizer & Metrics

```python
metrics = ['accuracy',
                tf.keras.metrics.AUC(),
               # tfa.metrics.CohenKappa(num_classes = 4),
               # tfa.metrics.F1Score(num_classes = 4),
                tf.keras.metrics.Precision(), 
                tf.keras.metrics.Recall()]
model.compile(loss =tf.keras.losses.CategoricalCrossentropy(), optimizer = 'adam', metrics = metrics)
print(model.summary())
```

### Early Stopping

```python
from tensorflow.keras.callbacks import EarlyStopping


early_stopping = EarlyStopping(
    monitor='val_loss',    
    patience=5,           
    restore_best_weights=True 
)
```

### Fitting the model

```python
history = model.fit(
    train_generator,
    epochs=30,
    validation_data=validation_generator,
    verbose=1,
    class_weight=class_weights,  # Include class weights here
    callbacks=[early_stopping]
)
```

### Classification Report

```python
test_steps_per_epoch = np.math.ceil(test_generator.samples / test_generator.batch_size)

predictions = model.predict(test_generator, steps = test_steps_per_epoch)

predicted_classes = np.argmax(predictions, axis=1)
```

```python
true_classes = test_generator.classes
class_labels = list(test_generator.class_indices.keys())
```

```python
report = sklearn.metrics.classification_report(true_classes, predicted_classes, target_names = class_labels)
print(report) 
```

### Confusion Matrix

```python
X_Label = ['CVN','DME','DRUSEN','NORMAL']
Y_Label = ['CVN','DME','DRUSEN','NORMAL']

cm = sklearn.metrics.confusion_matrix(true_classes, predicted_classes)
plt.figure(figsize=(8,8))
sns.heatmap(cm, fmt='.0f', cmap="crest", annot=True, linewidths=0.2, xticklabels=X_Label, yticklabels=Y_Label)
plt.title('confusion matrix')
plt.xlabel('predicted value')
plt.ylabel('Truth value')
plt.show()
print(sklearn.metrics.confusion_matrix(true_classes, predicted_classes))
```
